Overwriting covariate columns onto dataset


```python
#load in test
df_main = pd.read_csv("../../../data/finaldatasets/testdata/FixedRFdata (Copy).csv", na_values=["n/a", "missing", "-", "", "NA", "N/A"])


#load in covariate dataset
df_update = pd.read_csv("../../../data/finaldatasets/covariates/wheat_data_with_covariats.csv", na_values=["n/a", "missing", "-", "", "NA", "N/A"])
```

```python
print(df_main.columns.tolist())
print(df_update.columns.tolist())
```

```python
# Merge on 'id', keeping all rows from df_main
merged = df_main.merge(df_update, on="id", how="left", suffixes=("", "_new"))

# For each overlapping column, overwrite old with new (where new is not null)
for col in df_update.columns:
    if col != "id" and col in df_main.columns:
        merged[col] = merged[f"{col}_new"].combine_first(merged[col])
        merged.drop(columns=[f"{col}_new"], inplace=True)
```

```python

#sava dataset
df_main.to_csv("../../../data/finaldatasets/testdata/Overwrite.csv", index=False)
```

Adding GDP AND HDI index values to data

```python
#read in main data
df = pd.read_csv("../../../data/finaldatasets/testdata/FixedRFdatatojoin.csv", na_values=["n/a", "missing", "-", "", "NA", "N/A"])

# 1. Load NetCDF and your CSV
gdp_ds = xr.open_dataset("../../../data/finaldatasets/covariates/Covariates/GDP_per_capita_PPP_1990_2015_v2.nc")
hdi_ds = xr.open_dataset("../../../data/finaldatasets/covariates/Covariates/HDI_1990_2015_v2.nc")

# 2. Convert CSV columns to match NetCDF
df['year'] = df['Observation.period'].astype(float)
df['lat'] = df['Conversion.for.latitude']
df['lon'] = df['Conversion.for.longitude']

# 3. Interpolate NetCDF values
gdp_interp = gdp_ds['GDP_per_capita_PPP'].interp(
    time=xr.DataArray(df['year'], dims="points"),
    latitude=xr.DataArray(df['lat'], dims="points"),
    longitude=xr.DataArray(df['lon'], dims="points"),
    method="nearest"
)

hdi_interp = hdi_ds['HDI'].interp(
    time=xr.DataArray(df['year'], dims="points"),
    latitude=xr.DataArray(df['lat'], dims="points"),
    longitude=xr.DataArray(df['lon'], dims="points"),
    method="nearest"
)

# 4. Add to your dataframe
df["gdp"] = gdp_interp.values
df["hdi"] = hdi_interp.values

```

```python

#check if it worked
print(df.columns)

df.to_csv("../../../data/finaldatasets/testdata/RFgdpadded.csv", index=False)
```
